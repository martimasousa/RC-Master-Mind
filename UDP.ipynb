**Enviar uma mensagem usando um socket UDP do lado do cliente:**

1. Criar o socket UDP: Use socket(AF_INET, SOCK_DGRAM, 0).

2. Resolver o endereço do servidor: Use getaddrinfo() com o nome do servidor e a porta.

3. Enviar a mensagem: Use sendto() com o socket, a mensagem, o endereço do servidor e a porta.

4. Fechar o socket: Após o envio, use close() para liberar o socket.

**Receber uma mensagem usando um socket UDP do lado do servidor:**

1. Criar o socket UDP: Use socket(AF_INET, SOCK_DGRAM, 0).

2. Configurar o endereço para escutar

3. Associar o socket ao endereço: Use bind() para associar o socket ao endereço e porta.

4. Receber a mensagem: Use recvfrom() para receber a mensagem e o endereço do cliente.

5. Fechar o socket: Após o recebimento (e possível resposta), use close() para liberar o socket.

### int bind(int sockfd, const struct sockaddr *addr, socklen_t addrlen);
#### **WHAT DOES:** Associates a socket with a specific local address and port.

- **sockfd:** O descritor do socket a ser associado a um endereço específico. Este socket foi criado anteriormente com a função socket().
- **addr:** Um ponteiro para uma estrutura sockaddr (geralmente sockaddr_in para IPv4 ou sockaddr_in6 para IPv6) que contém o endereço (IP e porta) ao qual o socket será associado.
- **addrlen:** O tamanho da estrutura sockaddr que contém o endereço e a porta.

A função retorna:
- 0 em caso de sucesso.
- -1 em caso de erro, configurando a variável global errno com o código do erro.

### ssize_t sendto(int sockfd, const void *buf, size_t len, int flags, const struct sockaddr *dest_addr, socklen_t addrlen);
#### **WHAT DOES:** Sends a message to a socket

- **sockfd:** O descritor do socket usado para enviar os dados.
- **buf:** Ponteiro para o buffer que contém os dados a serem enviados.
- **len:** O tamanho (em bytes) dos dados contidos no buffer.
- **flags:** Permite definir opções adicionais para o envio. Usualmente é 0, indicando o comportamento padrão.
- **dest_addr:** Um ponteiro para uma estrutura sockaddr que especifica o endereço de destino.
- **addrlen:** O tamanho da estrutura apontada por dest_addr.

### ssize_t recvfrom(int sockfd, void *buf, size_t len, int flags, struct sockaddr *src_addr, socklen_t *addrlen);
#### **WHAT DOES:** Receives a messagem from a socket

- **sockfd:** O descritor do socket através do qual os dados serão recebidos. Esse socket deve ter sido criado e, no caso de UDP, deve ter sido associado a um endereço e porta com bind() (se o socket for usado em um servidor).
- **buf:** Um ponteiro para o buffer onde os dados recebidos serão armazenados. O buffer deve ser grande o suficiente para armazenar os dados que você espera receber.
- **len:** O tamanho máximo (em bytes) que o buffer pode armazenar. Ou seja, o número máximo de bytes a serem lidos da rede.
- **flags:** Usado para configurar opções adicionais para a operação de recebimento. Normalmente, pode ser 0, o que significa o comportamento padrão, mas pode ser usado para configurar opções como MSG_DONTWAIT ou MSG_PEEK (que alteram o comportamento do recebimento).
- **src_addr::** Um ponteiro para uma estrutura sockaddr (geralmente sockaddr_in para IPv4 ou sockaddr_in6 para IPv6) que será preenchida com o endereço de origem da mensagem. Isso inclui o IP e a porta do remetente.
- **addrlen:** Um ponteiro para o tamanho da estrutura src_addr. Antes de chamar recvfrom(), você deve definir addrlen como o tamanho da estrutura de endereço que está sendo passada. Após a chamada, addrlen conterá o tamanho da estrutura preenchida com as informações do remetente.
