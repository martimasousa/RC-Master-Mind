**Fluxo básico de um servidor TCP simples que envia uma mensagem para um servidor.**

1. Criar o socket TCP (socket(AF_INET, SOCK_STREAM, 0)).

2. Configurar o endereço do servidor (IP e porta).

3. Conectar ao servidor (connect()).

4. Enviar a mensagem para o servidor (send() ou write()).

5. Receber a resposta do servidor (opcional, com recv() ou read()).

6. Fechar o socket (close()).

**Fluxo básico de um servidor TCP simples que recebe uma mensagem de um cliente.**

1. Criar o socket TCP (socket(AF_INET, SOCK_STREAM, 0)).

2. Configurar o endereço do servidor (IP e porta).

3. Associar o socket ao endereço (bind()).

4. Colocar o socket em escuta (listen()).

5. Aceitar uma conexão do cliente (accept()).

6. Receber a mensagem do cliente (recv() ou read()).

7. Responder ao cliente (opcional, com send() ou write()).

8. Fechar os sockets (close()).

### int connect(int sockfd, struct sockaddr *addr, socklen_t addrlen);
#### **WHAT DOES:** Establishes a connection to a specified remote address (for a socket), typically used in client-server communication.
- **sockfd:** The socket file descriptor, returned by the socket() function. It represents the connection endpoint.
- **addr:** A pointer to a struct sockaddr that contains the address and port of the remote server to which the socket should connect. For IPv4 addresses, you would typically use struct sockddr_in.
- **addrlen:** The size of the addr structure, usually set using sizeof(struct sockaddr_in) for IPv4.

A função retorna:

- 0 em caso de sucesso, indicando que a conexão foi estabelecida com sucesso.
- -1 em caso de erro, e configura a variável global errno com o código de erro, indicando o motivo da falha (por exemplo, ECONNREFUSED para quando o servidor está indisponível).

### ssize_t write(int fd, const void *buf, size_t len);
or
### *int send(int sockfd, const void buf, size_t len, int flags);
### **WHAT DOES:** Sends data through a connected socket.
- **sockfd:** O descriptor do socket previamente criado com socket(). O socket deve estar conectado a um destino (como um servidor, no caso de TCP).
- **buf:** Um ponteiro para o buffer que contém os dados a serem enviados.
- **len:** O número de bytes a serem enviados a partir do buffer.
- **flags:** Flags de controle que modificam o comportamento da função (geralmente 0).

A função retorna:

- O número de bytes realmente enviados (pode ser menor que len se houver erro ou se os dados não puderem ser enviados totalmente).
-   -1 em caso de erro, e a variável global errno será configurada com o código do erro.

---

### *ssize_t read(int fd, void buf, size_t count);
or
### ssize_t recv(int sockfd, void *buf, size_t count, int flags);

#### **WHAT DOES:** Reads data from a socket (or file descriptor) into a buffer.
- **fd:** O descriptor do socket de onde os dados serão lidos. Para sockets conectados, o socket deve estar previamente conectado a um servidor (como no caso do TCP).
- **buf:** Um ponteiro para o buffer onde os dados lidos serão armazenados.
- **count:** O número máximo de bytes a serem lidos do socket.
- **flags:** Flags opcionais que controlam o comportamento da função.

A função retorna:

- O número de bytes lidos, que pode ser menor que o valor de count caso não haja mais dados disponíveis ou no caso de um erro.
- 0 se a conexão foi fechada de forma graciosa pelo outro lado.
- -1 em caso de erro, e a variável global errno será configurada com o código do erro.

### int listen(int sockfd, int backlog);
#### **WHAT DOES:** Coloca o socket em modo de escuta para conexões de entrada.
- **sockfd:** O descritor do socket previamente criado com socket() e associado a um endereço com bind().
- **backlog:** O número máximo de conexões pendentes na fila de espera que podem ser armazenadas antes de serem aceitas.

---

### **int accept(int sockfd, struct sockaddr addr, socklen_t addrlen);
#### **WHAT DOES:** Aceita uma conexão de entrada pendente em um socket passivo. É **BLOQUEANTE**
- **sockfd:** O descritor do socket que está em modo de escuta (configurado com listen()).
- **addr:** Um ponteiro para uma estrutura sockaddr que será preenchida com o endereço do cliente conectado (IP e porta).
- **addrlen:** Um ponteiro para um valor que inicialmente contém o tamanho da estrutura sockaddr. Após a execução, será atualizado para refletir o tamanho real dos dados do cliente.

### int sigaction( int signum, const struct sigaction *act, struct sigaction *oldact);
#### **WHAT DOES:** A função sigaction permite ao programa configurar como ele deve reagir a sinais específicos. Ela é usada para definir um manipulador de sinal personalizado, modificar o comportamento padrão de um sinal ou recuperar o manipulador de sinal anterior.
- **signum:** O número que identifica o sinal que você deseja manipular.
- **act:** Um ponteiro para uma estrutura struct sigaction que contém a ação a ser tomada quando o sinal ocorrer
- **oldact (opcional):** Um ponteiro para uma estrutura struct sigaction onde o manipulador de sinal anterior (se houver) será armazenado. Isso é útil se você quiser restaurar o comportamento original do sinal mais tarde.
