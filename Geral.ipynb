### int gethostname(char *name, size_t len);
#### **WHAT DOES:** Retrieves the hostname of the machine on which the program is running.

- **name:** Um ponteiro para o buffer onde o hostname será armazenado.
- **len:** O tamanho máximo que o buffer pode conter.

#### **Retorna:**

- 0 em caso de sucesso.
- -1 em caso de erro, configurando a variável global errno com o código do erro.

---

### int getnameinfo( const struct sockaddr *sa, socklen_t salen, char *host, socklen_t hostlen, char *service, socklen_t servicelen, int flags);
#### **WHAT DOES:** Resolves a socket address (IP address and port) into a human-readable hostname and service name.
- **sa:** Um ponteiro para a estrutura sockaddr que contém o endereço de rede (pode ser um endereço IPv4 ou IPv6). A estrutura que é passada depende do tipo de rede (IPv4 ou IPv6).
- **salen:** O tamanho da estrutura sockaddr que está sendo passada. Geralmente, isso é o tamanho da estrutura (como sizeof(struct sockaddr_in) para IPv4).
- **host:** Um ponteiro para um buffer onde o nome do host (nome de domínio ou endereço IP) será armazenado. Este buffer deve ser grande o suficiente para armazenar o nome do host resolvido.
- **hostlen:** O tamanho máximo do buffer host onde o nome do host será armazenado.
- **service:** Um ponteiro para um buffer onde o nome do serviço (nome da porta, como "http" ou "ftp") ou o número da porta será armazenado.
- **servicelen:** O tamanho máximo do buffer service onde o nome do serviço ou número da porta será armazenado.
- **flags:** Um conjunto de flags que podem modificar o comportamento da resolução, como NI_NUMERICHOST (retorna o endereço IP numérico) ou NI_NUMERICSERV (retorna o número da porta numérica).

#### **Retorna:**

- 0 em caso de sucesso.
- -1 em caso de erro, configurando a variável global errno com o código do erro.

### int getaddrinfo(const char *node, const char *service, const struct addrinfo *hints, struct addrinfo **res);
#### **WHAT DOES:** Resolves a hostname and service name into a list of socket address structures.

- **node:** Este é o nome do domínio que queremos resolver. Pode ser também um endereço IP, dependendo do caso.
- **service:** Define o serviço associado (ex.: número da porta para conexões TCP ou UDP). Quando é NULL, indica que não há um serviço específico associado. A resolução será apenas do endereço.
- **hints:** Um ponteiro para uma estrutura addrinfo que especifica preferências para a resolução de endereços (ex.: IPv4, IPv6, UDP, TCP, etc.).
- **res:** Um ponteiro para um ponteiro do tipo struct addrinfo. A função preencherá este ponteiro com uma lista ligada de resultados contendo as informações dos endereços resolvidos.
 
#### **Retorna:**

- 0: Em caso de sucesso.
- Gai_error: Um código de erro, que pode ser interpretado com a função gai_strerror() para obter uma descrição legível do erro.

### int socket(int domain, int type, int protocol);
#### **WHAT DOES:** Creates a socket.

- **domain:** Especifica o domínio de comunicação, ou seja, a família de endereços que o socket usará.
- **type:** Define o tipo de socket, ou seja, o tipo de comunicação que ocorrerá.
- **protocol:** Define o protocolo de transporte utilizado. Geralmente, você pode passar 0, permitindo que o sistema escolha automaticamente o protocolo adequado com base no tipo do socket.

#### **Retorna:**

- Descritor de socket (um número inteiro positivo) que é usado para referenciar o socket em operações subsequentes como bind(), send(), recv(), etc.
- -1 e configura a variável global errno para indicar o erro ocorrido.

### int select(int nfds, fd_set *readfds, fd_set *writefds, fd_set *exceptfds, struct timeval *timeout);
#### **WHAT DOES:** Monitora múltiplos descritores de arquivo, bloqueando até que um deles esteja pronto para operações de leitura, escrita ou ocorra uma exceção.

- **nfds:** O maior descritor de arquivo monitorado + 1. Usado para limitar a verificação.
- **readfds:** Um ponteiro para um conjunto de descritores monitorados para **leitura** (por exemplo, novos dados em um socket ou conexão pendente).
- **writefds:** Um ponteiro para um conjunto de descritores monitorados para **escrita** (por exemplo, capacidade de enviar dados).
- **exceptfds:** Um ponteiro para um conjunto de descritores monitorados para **condições excepcionais** (por exemplo, erro de out-of-band no TCP).
- **timeout:** Um ponteiro para um `struct timeval` que especifica o tempo máximo de espera:
  - `NULL` significa que a função bloqueia indefinidamente.
  - Um valor `struct timeval` define um tempo limite.

#### **Retorna:**
- O número de descritores prontos para operações solicitadas.
- 0 se o tempo limite expirar antes de qualquer atividade.
- -1 em caso de erro, configurando a variável global `errno`.
